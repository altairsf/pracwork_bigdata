```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('./dados/PicoWeb.csv', sep = ";", engine = "python")
```

```python
#Impressão das informações gerais sobre o conjunto de dados
df.info()
```

```python
#Impressão das primeiras linhas do arquivo
df.head()
```

```python
#Impressão das últimas linhas do arquivo
df.tail()
```

```python
#Variável com a cópia dos dados
df_copy = df.copy()
```

```python
#Substituição dos valores nulos da coluna 'Calories'
df_copy["Calories"] = df["Calories"].fillna(0)
```

```python
#Impressão do resultado da substituição acima
print(df_copy)
```

```python
#Substituição dos valores nulos da coluna "Date" por '1900/01/01'
df_copy["Date"] = df["Date"].fillna('1900/01/01')
```

```python
#Resultado da substituição
print(df_copy)
```

```python
#Transformação dos dados da coluna "Date" em datetime. Haverá um erro indicando que o valor '1900/01/01' não corresponde ao fomato de "date".
df_copy["Date"] = pd.to_datetime(df_copy["Date"])
```

```python
#Regularização dos dados acima por substituir o valor '1900/01/01' por 'NaN'. 
df_copy["Date"] = df_copy["Date"].replace("1900/01/01", 'NaN')
```

```python
#Utilização novamente do método 'to_datetime' para a transformação dos dados da coluna "Date". Agora, aparecerá um erro informando que o valor
#'20201226' não corresponde ao formato de "date".
df_copy["Date"] = pd.to_datetime(df_copy["Date"])
```

```python
#Impressão do resultado da substituição do valor '1900/01/01' na coluna 'Date' por 'NaN'
print (df_copy)
```

```python
#Regularização dos dados acima e da mensagem de erro. Transformação do valor "20201226", atualmente uma string, para o formato 'datetime'
df_copy["Date"] = df_copy["Date"].replace("20201226",pd.to_datetime("20201226", format="%Y%m%d"))
```

```python
#Utilização novamente do método 'to_datetime' para a transformação dos dados da coluna "Date"
df_copy["Date"] = pd.to_datetime(df_copy["Date"])
```

```python
#Impressão do conjunto de dados atual
print(df_copy)
```

```python
#Remoção dos registros com valores nulos. Após o tratamento, apenas a linha 22 tem esta condição.
df_copy = df_copy[df_copy["Date"].notna()].reset_index(drop=True)
```

```python
#Impressão final 
print(df_copy)
```
